```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("data/processed/clean_transactions.csv", parse_dates=["InvoiceDate"])
df.head()
```

```python
# Use the day after the latest transaction as the reference date.
# This prevents the most recent customer from having a recency of 0 days.
analysis_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)
print("Analysis date:", analysis_date)
```

```python
customer_features = df.groupby("CustomerID").agg(
    last_purchase_date=("InvoiceDate", "max"),
    first_purchase_date=("InvoiceDate", "min"),
    frequency=("InvoiceNo", "nunique"),
    monetary=("TotalAmount", "sum"),
    total_items=("Quantity", "sum"),
    product_variety=("ProductName", "nunique")
).reset_index()

customer_features["recency"] = (
    analysis_date - customer_features["last_purchase_date"]
).dt.days

customer_features["tenure_days"] = (
    analysis_date - customer_features["first_purchase_date"]
).dt.days

customer_features["avg_order_value"] = (
    customer_features["monetary"] / customer_features["frequency"]
)

customer_features.head()
```

```python
# Category spend shares
category_spend = df.pivot_table(
    index="CustomerID",
    columns="Category",
    values="TotalAmount",
    aggfunc="sum",
    fill_value=0
)

for category in ["Coffee Beans", "Brew Gear", "Accessories"]:
    if category in category_spend.columns:
        customer_features[f"{category.lower().replace(' ', '_')}_share"] = (
            customer_features["CustomerID"]
            .map(category_spend[category])
            .fillna(0)
            / customer_features["monetary"]
        )

customer_features.head()
```

```python
features_for_clustering = [
    "recency",
    "frequency",
    "monetary",
    "avg_order_value",
    "tenure_days",
    "coffee_beans_share",
    "brew_gear_share",
    "accessories_share"
]

# Keep only features that exist in the dataset
features_for_clustering = [
    feature for feature in features_for_clustering
    if feature in customer_features.columns
]

customer_features[features_for_clustering].describe()
```

```python
# Reduce skewness in monetary and frequency-related features
skewed_features = ["monetary", "frequency", "avg_order_value", "tenure_days"]

for feature in skewed_features:
    if feature in customer_features.columns:
        customer_features[f"{feature}_log"] = np.log1p(customer_features[feature])

# Scale features for distance-based clustering
scaled_features = [
    "recency",
    "frequency_log",
    "monetary_log",
    "avg_order_value_log",
    "tenure_days_log",
    "coffee_beans_share",
    "brew_gear_share",
    "accessories_share"
]

scaled_features = [
    feature for feature in scaled_features
    if feature in customer_features.columns
]

scaler = StandardScaler()
customer_features_scaled = scaler.fit_transform(customer_features[scaled_features])

scaled_df = pd.DataFrame(
    customer_features_scaled,
    columns=scaled_features
)

scaled_df["CustomerID"] = customer_features["CustomerID"]
scaled_df.to_csv("data/processed/customer_features_scaled.csv", index=False)
customer_features.to_csv("data/processed/customer_features.csv", index=False)

print("Customer features saved.")
customer_features.head()
```
