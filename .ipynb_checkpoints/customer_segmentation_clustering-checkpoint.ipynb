```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.decomposition import PCA
```

```python
scaled_df = pd.read_csv("data/processed/customer_features_scaled.csv")
customer_features = pd.read_csv("data/processed/customer_features.csv")
```

```python
X = scaled_df.drop(columns=["CustomerID"])

print("Feature matrix shape:", X.shape)
X.head()
```

```python
def evaluate_model(model_name, labels, X):
    return {
        "Model": model_name,
        "Clusters": len(np.unique(labels)),
        "Silhouette Score": silhouette_score(X, labels),
        "Davies-Bouldin Index": davies_bouldin_score(X, labels)
    }

results = []

# K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans_labels = kmeans.fit_predict(X)
results.append(evaluate_model("K-Means", kmeans_labels, X))

# Agglomerative Clustering
agglomerative = AgglomerativeClustering(n_clusters=4)
agglomerative_labels = agglomerative.fit_predict(X)
results.append(evaluate_model("Agglomerative Clustering", agglomerative_labels, X))

# Gaussian Mixture Model
gmm = GaussianMixture(n_components=4, random_state=42)
gmm_labels = gmm.fit_predict(X)
results.append(evaluate_model("Gaussian Mixture Model", gmm_labels, X))

model_comparison = pd.DataFrame(results)
model_comparison
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

sns.barplot(
    data=model_comparison,
    x="Model",
    y="Silhouette Score",
    ax=axes[0],
    hue="Model",
    palette="viridis"
)
axes[0].set_title("Silhouette Score by Model")
axes[0].tick_params(axis="x", rotation=20)

sns.barplot(
    data=model_comparison,
    x="Model",
    y="Davies-Bouldin Index",
    ax=axes[1],
    hue="Model",
    palette="mako"
)
axes[1].set_title("Davies-Bouldin Index by Model")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.savefig("Outputs/Figures/clustering_model_comparison.png", bbox_inches="tight")
plt.show()
```

```python
pca = PCA(n_components=2, random_state=42)
pca_components = pca.fit_transform(X)

pca_df = pd.DataFrame(pca_components, columns=["PC1", "PC2"])
pca_df["KMeans_Cluster"] = kmeans_labels
pca_df["Agglomerative_Cluster"] = agglomerative_labels
pca_df["GMM_Cluster"] = gmm_labels

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="KMeans_Cluster", palette="tab10", s=20, ax=axes[0])
axes[0].set_title("K-Means Clusters")

sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Agglomerative_Cluster", palette="tab10", s=20, ax=axes[1])
axes[1].set_title("Agglomerative Clusters")

sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="GMM_Cluster", palette="tab10", s=20, ax=axes[2])
axes[2].set_title("Gaussian Mixture Clusters")

plt.tight_layout()
plt.savefig("Outputs/Figures/clustering_models_pca.png", bbox_inches="tight")
plt.show()
```
